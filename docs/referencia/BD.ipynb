```r


#install.packages(c("googledrive", "readxl", "dplyr"))


library(googledrive)
library(readxl)
library(dplyr)
drive_auth()


carpeta <- drive_get("Multivariados_Reto")

carpeta

archivos <- drive_ls(carpeta)

archivos$name

archivos <- archivos[
  grepl("\\.xlsx$", archivos$name, ignore.case = TRUE),
]

archivos$name
```

```r
library(googledrive)
library(readxl)
library(dplyr)

# Base final vacía
BD <- data.frame()

# Recorrer los archivos
for (i in 1:nrow(archivos)) {

  nombre_archivo <- archivos$name[i]

  # Extraer el año del nombre
  año <- sub(".*(20[0-9]{2}).*", "\\1", nombre_archivo)
  archivo <- tempfile(fileext = ".xlsx")

  drive_download(
    file = archivos[i, ],
    path = archivo,
    overwrite = TRUE
  )

  hojas <- excel_sheets(archivo)

  for (hoja in hojas) {

    datos <- read_excel(
      archivo,
      sheet = hoja
    )




    if ("date" %in% names(datos)) {
      names(datos)[names(datos) == "date"] <- "Fecha y hora"
    }

    if ("Date" %in% names(datos)) {
      names(datos)[names(datos) == "Date"] <- "Fecha y hora"
    }

    if ("fecha" %in% names(datos)) {
      names(datos)[names(datos) == "fecha"] <- "Fecha y hora"
    }



    datos$Año <- año
    datos$Planta <- hoja

    BD <- bind_rows(BD, datos)
  }

  unlink(archivo)
}

dim(BD)

names(BD)

head(BD)
```

```r
colSums(is.na(BD))
```

```r
str(BD)
summary(BD)
variables = c("CO", "NO", "NO2", "NOX", "O3", "PM10",
              "PM2.5", "PRS", "RAINF", "RH", "SO2",
              "SR", "TOUT", "WSR", "WDR")

for (variable in variables) {

  grafica = ggplot(BD, aes(x = factor(Año), y = .data[[variable]])) +
    geom_boxplot() +
    labs(
      title = paste(variable, "por año"),
      x = "Año",
      y = variable
    ) +
    theme_minimal()

  print(grafica)
}

```
